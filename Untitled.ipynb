```python
import numpy as np

print(np.__version__)
```

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6],
])

print(a.ndim)   # 2：数组有两个维度
print(a.shape)  # (2, 3)：第0维长度2，第1维长度3
print(a.size)   # 6：元素总数
print(a.dtype)  # 整数类型，例如 int64
```

```python
a.ndim == len(a.shape)
a.size == np.prod(a.shape)
```
